```python
#Read all datasets
import pandas as pd
from pathlib import Path

folder = Path("files (1)")

datasets = {}

for file in folder.glob("*.csv"):
    df = pd.read_csv(file)
    datasets[file.name] = df

    print(file.name, df.shape)
```

```python
for filename, df in datasets.items():
    print("\nFile:", filename)
    print(df.columns.tolist())
```

```python
#Validate schema consistency
expected_schema = {
    "customers.csv": {
        "customer_id": "integer",
        "customer_name": "string",
        "email": "string",
        "city": "string",
        "signup_date": "date"
    },
    "orders.csv": {
        "order_id": "integer",
        "customer_id": "integer",
        "product_id": "integer",
        "order_date": "date",
        "quantity": "integer",
        "unit_price": "numeric",
        "payment_status": "string"
    },
    "products.csv": {
        "product_id": "integer",
        "product_name": "string",
        "category": "string",
        "supplier_id": "integer",
        "cost_price": "numeric",
        "selling_price": "numeric"
    },
    "clickstream.csv": {
        "event_id": "integer",
        "customer_id": "integer",
        "event_type": "string",
        "page_url": "string",
        "event_timestamp": "date",
        "device_type": "string"
    }
}

for filename, df in datasets.items():
    print("\nDataset:", filename)

    expected = expected_schema[filename]

    missing = set(expected) - set(df.columns)
    extra = set(df.columns) - set(expected)

    print("Missing columns:", missing)
    print("Extra columns:", extra)

    for col, dtype in expected.items():
        if col not in df.columns:
            continue

        values = df[col]
        present = values.notna()

        if dtype in ("integer", "numeric"):
            parsed = pd.to_numeric(values, errors="coerce")
            invalid = present & parsed.isna()

            if dtype == "integer":
                invalid |= parsed.notna() & (parsed % 1 != 0)

        elif dtype == "date":
            parsed = pd.to_datetime(
                values,
                format="mixed",
                dayfirst=True,
                errors="coerce",
                utc=True
            )
            invalid = present & parsed.isna()

        else:
            invalid = present & ~values.map(
                lambda x: isinstance(x, str)
            )

        print(col, "- invalid types:", invalid.sum())
```

```python
#Detect duplicate customer
customers = datasets["customers.csv"]

# Duplicate customer IDs
duplicate_ids = customers[
    customers["customer_id"].duplicated(keep=False)
]

# Duplicate emails
normalized_email = (
    customers["email"]
    .astype("string")
    .str.strip()
    .str.lower()
)

duplicate_emails = customers[
    normalized_email.notna()
    & normalized_email.duplicated(keep=False)
]

print("Duplicate ID records:", len(duplicate_ids))
print("Duplicate email records:", len(duplicate_emails))

print(duplicate_ids)
print(duplicate_emails)
```

```python
customers = datasets["customers.csv"]
orders = datasets["orders.csv"]
products = datasets["products.csv"]
clickstream = datasets["clickstream.csv"]
```

```python
#Identify invalid records
# Customers missing required information
invalid_customers = customers[
    customers["customer_id"].isna()
    | customers["customer_name"].isna()
    | customers["email"].isna()
]

# Invalid orders
invalid_orders = orders[
    orders["order_id"].isna()
    | orders["customer_id"].isna()
    | orders["product_id"].isna()
    | (orders["quantity"] <= 0)
    | (orders["unit_price"] < 0)
    | ~orders["customer_id"].isin(customers["customer_id"])
    | ~orders["product_id"].isin(products["product_id"])
]

# Invalid products
invalid_products = products[
    products["product_id"].isna()
    | (products["cost_price"] < 0)
    | (products["selling_price"] < 0)
]

# Invalid clickstream records
invalid_clickstream = clickstream[
    clickstream["event_id"].isna()
    | clickstream["event_timestamp"].isna()
]

print("Invalid customers:", len(invalid_customers))
print("Invalid orders:", len(invalid_orders))
print("Invalid products:", len(invalid_products))
print("Invalid clickstream:", len(invalid_clickstream))
```

```python
#Q2. Data Cleaning & Transformation
#Making copies so the original datasets remain unchanged
customers_clean = customers.copy()
orders_clean = orders.copy()
products_clean = products.copy()
clickstream_clean = clickstream.copy()
```

```python
#Standardize city names
customers_clean["city"] = (
    customers_clean["city"]
    .astype("string")
    .str.strip()
    .str.lower()
    .str.title()
)

print(customers_clean["city"].value_counts())
```

```python
#Convert timestamps correctly
clickstream_clean["event_timestamp"] = pd.to_datetime(
    clickstream_clean["event_timestamp"],
    format="mixed",
    dayfirst=True,
    errors="coerce",
    utc=True
)

customers_clean["signup_date"] = pd.to_datetime(
    customers_clean["signup_date"],
    errors="coerce"
)

orders_clean["order_date"] = pd.to_datetime(
    orders_clean["order_date"],
    errors="coerce"
)

print(clickstream_clean["event_timestamp"].isna().sum())
```

```python
#Handling missing values
for name, df in {
    "customers": customers_clean,
    "orders": orders_clean,
    "products": products_clean,
    "clickstream": clickstream_clean
}.items():
    print("\n", name)
    print(df.isna().sum())
```

```python
clickstream_clean = clickstream.copy()
# Customer ID: keep missing values (anonymous users)
# No change needed

# Page URL: keep missing values
# Exclude these records when analyzing most visited pages

# Device type: replace missing values with "Unknown"
clickstream_clean["device_type"] = (
    clickstream_clean["device_type"].fillna("Unknown")
)
```

```python
print(clickstream_clean.isnull().sum())
```

```python
bad_customers = (
    customers_clean["customer_id"].isna()
    | customers_clean["customer_name"].isna()
    | customers_clean["email"].isna()
)

bad_orders = (
    orders_clean["order_id"].isna()
    | orders_clean["customer_id"].isna()
    | orders_clean["product_id"].isna()
    | orders_clean["order_date"].isna()
    | (orders_clean["quantity"] <= 0)
    | (orders_clean["unit_price"] < 0)
)

bad_products = (
    products_clean["product_id"].isna()
    | (products_clean["cost_price"] < 0)
    | (products_clean["selling_price"] < 0)
)

bad_clickstream = (
    clickstream_clean["event_id"].isna()
    | clickstream_clean["event_timestamp"].isna()
)

# Keep rejected rows
rejected_clickstream = clickstream_clean[bad_clickstream].copy()

# Keep valid rows
customers_clean = customers_clean[~bad_customers].copy()
orders_clean = orders_clean[~bad_orders].copy()
products_clean = products_clean[~bad_products].copy()
clickstream_clean = clickstream_clean[~bad_clickstream].copy()
```

```python
#saving cleaned values
customers_clean.to_csv("customers_clean.csv", index=False)
orders_clean.to_csv("orders_clean.csv", index=False)
products_clean.to_csv("products_clean.csv", index=False)
clickstream_clean.to_csv("clickstream_clean.csv", index=False)

rejected_clickstream.to_csv(
    "clickstream_rejected.csv", index=False
)
```

```python
# Q3 - Task 2: Calculate Session Counts

# Create a copy of the cleaned clickstream dataset
cs = clickstream_clean.copy()

# Exclude records without customer ID or timestamp
cs = cs.dropna(subset=["customer_id", "event_timestamp"])

# Ensure timestamps are datetime values
cs["event_timestamp"] = pd.to_datetime(
    cs["event_timestamp"],
    format="mixed",
    errors="coerce",
    utc=True
)

cs = cs.dropna(subset=["event_timestamp"])

# Sort events by customer and timestamp
cs = cs.sort_values(["customer_id", "event_timestamp"])

# Calculate time gap between consecutive events
cs["time_diff"] = (
    cs.groupby("customer_id")["event_timestamp"].diff()
)

# Identify new sessions using 30-minute inactivity rule
cs["new_session"] = (
    cs["time_diff"].isna()
    | (cs["time_diff"] > pd.Timedelta(minutes=30))
)

# Assign session numbers for each customer
cs["session_number"] = (
    cs.groupby("customer_id")["new_session"].cumsum()
)

# Calculate total estimated sessions
total_sessions = int(cs["new_session"].sum())

print("Total Estimated Sessions:", total_sessions)

# Calculate sessions per customer
sessions_per_customer = (
    cs.groupby("customer_id")["session_number"].max()
)

print("\nSessions Per Customer:")
print(sessions_per_customer.head(10))
```
